# Chapter 32 — Statistical Decision Theory: Loss, Utility and the Value of Information

## WHAT
This is a chapter-specific, executable synthetic-data laboratory addressing **asymmetric decision loss and information value**. It illustrates the manuscript's proposition: Evidence describes uncertainty; a decision also requires objectives, costs, benefits, and constraints. Two decision-makers may rationally choose different actions from the same probability estimate because their loss functions differ.

## WHY
The experiment makes an analytical or inferential distinction visible under controlled assumptions. The synthetic observational units have known generating mechanisms. This is **not** an analysis of actual people, institutions or historical records, and does not establish general scientific validity.
